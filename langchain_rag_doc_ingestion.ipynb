```python
# configuration
docs_dir_path = "/mnt/d/ANSAL-3/ML Model Projects/Langchain-RAG/docs"
vector_db_path = "/mnt/d/ANSAL-3/ML Model Projects/Langchain-RAG/vector_db"
collection_name = "document_collection"
```

```python
# loading the embedding model
embedding = HuggingFaceEmbeddings()
```

```python
# directory loader
loader = DirectoryLoader(
    path=docs_dir_path,
    glob="./*.pdf",
    loader_cls=UnstructuredFileLoader
)

# single file can also be loaded with UnstructuredFileLoader
```

```python
# load the documents
documents = loader.load()
```

```python
print(type(documents))
```

```python
len(documents)
```

```python
# initializing the text splitter
text_splitter = CharacterTextSplitter(
    chunk_size=2000,
    chunk_overlap=500
)
```

```python
# splitting the text into smaller chunks
text_chunks = text_splitter.split_documents(documents)
```

```python
len(text_chunks)
```

```python
# creating the vector store
vector_store = Chroma.from_documents(
    documents=text_chunks,
    embedding=embedding,
    persist_directory=vector_db_path,
    collection_name=collection_name
)
```
